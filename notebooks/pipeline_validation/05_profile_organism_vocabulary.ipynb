```python
%sql
select
    cohort_id,
    organism,
    count(*) as culture_organism_rows,
    count(distinct patient_key) as patients
from workspace.micro_dev.int_cohort_culture_organism
group by
    cohort_id,
    organism
order by
    cohort_id,
    culture_organism_rows desc;
```

### Frequency by cohort

This makes it easy to see which source labels account for most of the volume in each cohort and which mappings are worth reviewing first.

```python
%sql
select
    organism,
    count(*) as culture_organism_rows,
    count(distinct culture_key) as culture_orders,
    count(distinct patient_key) as patients
from workspace.micro_dev.int_cohort_culture_organism
group by organism
order by culture_organism_rows desc;
```

### Overall frequency

The same organism can appear under more than one source label, so this is still a source-level profile rather than a reporting taxonomy.

```python
%sql
select
    organism,
    count(*) as culture_organism_rows
from workspace.micro_dev.int_cohort_culture_organism
where organism like '%SPECIES%'
   or organism like '%GROUP%'
   or organism like '%COMPLEX%'
   or organism like '%MUCOID%'
   or organism like '%CF%'
group by organism
order by culture_organism_rows desc;
```

### Labels that need terminology review

I pulled labels containing terms such as `SPECIES`, `GROUP`, `COMPLEX`, phenotype qualifiers, and morphotype language. These are the labels where a simple text normalization can easily change the biological meaning.

```python
%sql
select
    cohort_id,
    count(distinct organism) as distinct_source_organism_labels
from workspace.micro_dev.int_cohort_culture_organism
group by cohort_id
order by cohort_id;
```

### Vocabulary by cohort

Blood has 238 distinct source labels, respiratory 167, and urine 142.

### Takeaway

The mapping layer needs to be conservative. Formatting variants can be standardized, but species, complexes, groups, and phenotype-qualified names should only be combined when I have an explicit reporting rule for them.
